```python
# install first:
# pip install pymupdf

import pymupdf as fitz
import os
import re
from pathlib import Path

# =========================
# CONFIG
# =========================
PDF_PATH = "/content/drive/MyDrive/Morgan Housel - Same as Ever_ A Guide to What Neve_241209_120403.pdf"          # your pdf file
OUTPUT_DIR = "chunks"
WORDS_PER_CHUNK = 1500

# =========================
# CREATE OUTPUT DIRECTORY
# =========================
Path(OUTPUT_DIR).mkdir(parents=True, exist_ok=True)

# =========================
# READ PDF
# =========================
doc = fitz.open(PDF_PATH)

all_text = []

for page in doc:
    text = page.get_text()
    all_text.append(text)

doc.close()

full_text = "\n".join(all_text)

# =========================
# CLEAN TEXT
# =========================
full_text = re.sub(r'\s+', ' ', full_text).strip()

# =========================
# SPLIT INTO WORDS
# =========================
words = full_text.split()

# =========================
# CHUNKING
# =========================
chunks = []

for i in range(0, len(words), WORDS_PER_CHUNK):
    chunk_words = words[i:i + WORDS_PER_CHUNK]
    chunk_text = " ".join(chunk_words)
    chunks.append(chunk_text)

# =========================
# SAVE CHUNKS
# =========================
for idx, chunk in enumerate(chunks, start=1):
    filename = f"part{idx:02d}.txt"
    filepath = os.path.join(OUTPUT_DIR, filename)

    with open(filepath, "w", encoding="utf-8") as f:
        f.write(chunk)

    print(f"Saved: {filepath}")

print(f"\nDone. Total chunks created: {len(chunks)}")
```

```python
import shutil

# zip the folder
shutil.make_archive("chunks", "zip", "chunks")

print("ZIP created: chunks.zip")
```
